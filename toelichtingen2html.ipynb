```python
from bs4 import BeautifulSoup
toelichtingen = {}
flnms = {}
fllst = [f for f in os.listdir(basedir) if os.path.splitext(f)[-1]=='.html']
for item in fllst:
    nm = os.path.splitext(item)[0]
    knm = nm.split('(')[0].strip()
    lnm = f'{os.path.join(basedir,item)}'
    flnms[nm] = lnm
    with open(os.path.join(basedir,item),encoding='utf8') as infl:
        # print (f'parsing {infl}')
        htmlsource = BeautifulSoup(infl, 'lxml')
        # print(htmlsource)
        toelichting = htmlsource('body')[0].renderContents().strip()
        toelichtingen[knm] = toelichting
        # clgs[nm] = toelichting
        # clgs[knm] = toelichting 
```

```python
def kill_trema(x): #
    return x.replace('ë','ë')
```

```python
colleges.college = colleges.college.apply(kill_trema)
```

```python
cnms = [nm.split('(')[0].strip() for nm in colleges.college.unique()]
cnms
```

```python
import os
from bs4 import BeautifulSoup
clgs = {}
for toedir in [d for d in con['toe_dirs'] if 'Republiek' not in d]:
    fllst = os.listdir(toedir)
    for item in fllst:
        item = kill_trema(item)
        nm = os.path.splitext(item)[0]
        knm = nm.split('(')[0].strip()
        if nm in cnms:
            naam = nm
        elif knm in cnms:
            naam = knm
        else:
            print(nm)
            # if nm in ['Departement van Publiek Onderwijs, Nationale Nijverheid en Koloniën (1818-1824)',
            #           'Departement van Koloniën (1814-1861)',
            #           'Hoog Gerechtshof (1813-1838)']: #wtf
            #     continue
            # else:
            break
        lnm = f'{os.path.join(basedir,item)}'
        flnms[nm] = lnm
        # print(nm)
        with open(os.path.join(toedir,item),encoding='latin1') as infl:
            htmlsource = BeautifulSoup(infl, 'html.parser')
            toelichting = htmlsource('body')[0].renderContents().strip()

            clgs[nm] = {'filename':lnm,'dbnaam':nm, 'naam':naam,'toelichting':toelichting}
            # clgs[knm] = toelichting 
```
